4.1 - The Normal-Gamma Conjugate Family
* It is common in practice to have data where the variability of observations is unknown
* Bayesian inference for two or more parameters is not different from the case with one parameter
* Bayes theorem for the posterior distribution takes the form of $p(\mu,\sigma^2 \mid y_1,\ldots,y_n) = \dfrac{p(y_1,\ldots,y_n \mid \mu,\sigma^2) \times p(\mu,\sigma^2)}{\text{normalizing constant}}$
* Conjugate priors are appealing as there are nice expressions for updating the prior to obtain the posterior distrbution using summaries of the data
* We assume that the data are a random sample of size n from a normal population with mean mu and variance sigma squared
* The posterior distribution should factor as the product of two conjugate distributions
