3.1 - Bayesian Decision Making
* The posterior distribution is the basis of any inference, since it integrates prior opinions and knowledge with the new information provided by the data
* Credible interval - a range of likely values for a parameter

3.2 - Loss Functions
* If you are *linear* in your beliefs, the optimal one number estimate is the *median* of the posterior distribution
* In the case that loss is proportional to the *quadratic* error, the optimal one number estimate is the *mean* of the posterior distribution
* If loss is *binary* you want to maximize the expected gain, meaning the optimal one number estimate is the *mode* of the posterior distribution

3.3 - Working with Loss Functions
* The loss function for (0/1) loss is $L_0 = \sum_i L_{0,i}(0, g)$
* The loss function for linear loss is $L_1(g) = \sum_i |x_i - g|$ 
* If the loss function is squared, the total loss for a guess is the sum of squared differences between that guess and each value in the posterior

3.4 - Minimizing Expected Loss for Hypothesis Testing
* If we have two competing hypotehsis (H1 and H2) we get 
    * P(H1 is true | data) = posterior probability of H1
    * P(H2 is true | data) = posterior probability of H2
    * Decision criteria becomes reject H1 if P(H1 is true | data) < P(H2 is ture | data)
* HIV example
    * 2 possible decisions
        * d1: choose H1 and decide that the patient does not have HIV
        * d2: choose H2 and decide that the patient has HIV
    * If d = d1 and we are right there is no loss and if d = d2 and we are right there is no loss
    * If d = d1 and we are wrong that is a false negative. Consequences are no treatment and premature death
    * If d = d2 and we are wrong that is a false positive. Consequences are distress.
        * It is important to not that being wrong does not always have the same consequences.
* Bayesian methodologies allow for the integration of potential losses into the decision making framework

3.5 - Posterior Probabilities of Hypotheses and Bayes Factors
* Prior odds - ratio of prior probabilites of hypotheses
* If there are two competing hypotheses being considered, then to prior odds of H1 to H2 can be defined as O[H1 : H2]
    * $O[H_1 : H_2] = \dfrac{P(H_1)}{P(H_2)}$
* The posterior odds is the ratio of two posterior probabilites of hypotheses
    * $PO[H_1 : H_2] = \dfrac{P(H_1 | \text{data})}{P(H_2 | \text{data})}$
* In a discrete case, the Bayes factor is the ratio of the likelihoods of the observed data under the two hypotheses
    * $BF[H_1 : H_2] = \dfrac{P(\text{data} | H_1)}{P(\text{data} | H_2)}$
* And in a continuous case, the BF if the ratio of the marginal likelihoods
    * $BF[H_1 : H_2] = \dfrac{\int P(\text{data} | \theta, H_1) d\theta}{\int P(\text{data} | \theta, H_2) d\theta}$
